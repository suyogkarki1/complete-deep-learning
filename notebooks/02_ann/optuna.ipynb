```python
import requests

url = "https://tcga.xenahubs.net/download/TCGA.PANCAN.sampleMap/EB++AdjustPANCAN_IlluminaHiSeq_RNASeqV2.geneExp.xena.gz"

response = requests.get(
    url,
    stream=True,
    headers={"User-Agent": "Mozilla/5.0"}
)

print(response.status_code)

response.raise_for_status()

with open("tcga_expression.xena.gz", "wb") as file:
    for chunk in response.iter_content(chunk_size=1024 * 1024):
        if chunk:
            file.write(chunk)

print("Download complete.")
```
