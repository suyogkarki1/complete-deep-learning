FOR pytorch we need to make all the data into tensor 

```python
import torch
```

```python
X_train_tensor=torch.Tensor(X_train)
X_test_tensor=torch.Tensor(X_test)
y_train_tensor=torch.Tensor(y_train)
y_test_tensor=torch.Tensor(y_test)
```

```python
X_train_tensor.shape
```

```python
type(X_train_tensor)
```

```python
class MyANN():
    def __init__(self,X):
        self.weights=torch.rand(X.shape[1],1,dtype=torch.float32,requires_grad=True)
        self.bias=torch.zeros(1,dtype=torch.float32,requires_grad=True)
        
    def forward(self,X):
        z=torch.matmul(X,self.weights)+self.bias
        y_pred=torch.sigmoid(z)
        return y_pred
    
    def loss_function(self,y_pred,y):
        #clamp prediction to avoid log(0)
        epsilon=1e-7
        y_pred=torch.clamp(y_pred,epsilon,1-epsilon)
        
        #loss function 
        ## formula of Binary cross entropy 
        loss=-(y_train_tensor * torch.log(y_pred) + (1-y_train_tensor)*torch.log(1-y_pred)).mean()
        return loss
    
        
    
```

```python
learning_rate=0.1
epochs=50
```

```python
model=MyANN(X_train_tensor)

for epochs in range(epochs):
    
    #forward pass
    y_pred=model.forward(X_train_tensor)
    
    #loss calculate
    loss=model.loss_function(y_pred,y_train_tensor)
    
    
    #backward
    loss.backward()
    
    #parameter updates wnew=wold-lr del l /del w old
    with torch.no_grad():
        model.weights-=learning_rate * model.weights.grad
        model.bias-=learning_rate * model.bias.grad
        
    #zero gradients for every iteration we make the gradient zero 
    model.weights.grad.zero_()
    model.bias.grad.zero_()
    
    #Print loss in each epochs
    
    print(f'epoch:{epochs+1},loss:{loss.item()}')
    
    
```

#evalution 

```python
with torch.no_grad():  #gradient tracking off 
    y_pred=model.forward(X_test_tensor)
    y_pred=(y_pred >0.9).float()
    accuracy=(y_pred==y_test_tensor).float().mean()
    print(f'Acuracy:{accuracy.item()}')
```
