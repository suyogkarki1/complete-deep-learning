```python
import torch
```

### Autograd automatic differentiation for tensor operation 
### its enable gradient component 

```python
x=torch.tensor(2.0, requires_grad=True)
#scaler
```

```python
y = 3*x**5 - 4*x**4 + 2*x**3 - 7*x**2 + 5*x - 1
```

```python
x
```

```python
y
```

1st derivates

```python

y.backward(create_graph=True)
first_derivates=x.grad
print('frist derivates',first_derivates)
```

2 derivates

```python
x.grad =None
first_derivates.backward(create_graph=True)
second_derivates=x.grad
print('second Derivates',second_derivates)
```

3rd derivates

```python
x.grad=None
second_derivates.backward(create_graph=True)
third_derivates=x.grad
print('third derivates',third_derivates)
```

fourth derivates

```python
x.grad=None
third_derivates.backward(create_graph=True)
fourth_derivates=x.grad
print('FOurth Derivates',fourth_derivates)
```

fifth Derivates

```python
x.grad=None
fourth_derivates.backward(create_graph=True)
fifth_derivates=x.grad
print('FOurth Derivates',fifth_derivates)
```

Sixth Derivates

```python
x.grad=None
fifth_derivates.backward(create_graph=True)
sixth_derivates=x.grad
print('FOurth Derivates',sixth_derivates)
```
