# Matrix Operation 

```python
f=torch.randint(size=(2,3),low=0 , high=10)
g=torch.randint(size=(3,2),low=0 , high=10)
print(f)
print(g)
```

```python
#Matrix multiplication 
torch.matmul(f,g)
```

```python
#dot product 

vector1=torch.tensor([2,3])
vector2=torch.tensor([2,3])
```

```python
torch.dot(vector1,vector2)
```

```python
h=torch.randint(size=(3,3),high=10,low=0,dtype=torch.float32)
```

```python
#determinant  only on square matrix 
torch.det(h)
```

```python
torch.inverse(h)
```
